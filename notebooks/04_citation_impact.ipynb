# RQ2 — Observed citation impact
This notebook reads verified Phase 4 summaries only. Citation counts are skewed and age-dependent; they are not scientific quality. FWCI complete cases exclude every sampled preprint. Methods live in `src/ai_research_trends/analysis/citation_impact.py` and `statistical_tests.py`.

```python
from pathlib import Path
import pandas as pd
from IPython.display import Image, Markdown, display
root = Path.cwd()
if not (root / 'pyproject.toml').exists(): root = root.parent
assert (root / 'pyproject.toml').exists()
out = root / 'data/analysis/phase4_stratified_20261019T155909Z_v3'
assert (out / 'manifest.json').exists(), 'Run openalex-analyze-phase4 first'
citations = pd.read_csv(out / 'citation_descriptives.csv')
fwci = pd.read_csv(out / 'fwci_descriptives.csv')
tests = pd.read_csv(out / 'statistical_test_results.csv')
display(citations.query("group_by == 'all' or group_by == 'primary_subfield_id'").head(10))
display(fwci.query("group_by == 'all_eligible' or group_by == 'work_type'"))
display(tests)
```

```python
display(Image(filename=str(out / 'figures/rq2_citation_distribution.png'), width=750))
display(Image(filename=str(out / 'figures/rq2_citation_age.png'), width=750))
display(Image(filename=str(out / 'figures/rq2_fwci_by_subfield.png'), width=750))
all_cites = citations.query("group_by == 'all' and metric == 'cited_by_count'").iloc[0]
all_fwci = fwci.query("group_by == 'all_eligible'").iloc[0]
display(Markdown(f'**Finding:** Median citations are {all_cites["median"]:.1f} versus mean {all_cites["mean"]:.2f} across {int(all_cites.n):,} sampled works. FWCI is available for {int(all_fwci.n):,} article/review works; preprints are not represented. The reported Spearman test is exploratory conditional-sample evidence.'))
```
