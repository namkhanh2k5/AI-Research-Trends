```python
from pathlib import Path
import pandas as pd
from IPython.display import Image, Markdown, display
root = Path.cwd()
if not (root / 'pyproject.toml').exists(): root = root.parent
assert (root / 'pyproject.toml').exists()
out = root / 'data/analysis/phase4_stratified_20261019T155909Z_v3'
assert (out / 'manifest.json').exists(), 'Run openalex-analyze-phase4 first'
growth = pd.read_csv(out / 'publication_trends.csv')
native = pd.read_csv(out / 'native_topic_trends.csv')
emerging = pd.read_csv(out / 'emerging_topic_sample_trends.csv')
display(growth[['year','work_count','absolute_yoy_change','yoy_growth_percent','index_2020_100']])
display(native[['series','year','work_count','share_of_primary_percent']].head(10))
display(emerging[['category','year','matched_sample_works','sample_year_denominator','abstract_available_works']].head(10))
```
